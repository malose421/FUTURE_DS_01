```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
meta = pd.read_csv("ee_meta.csv")

profile = pd.read_csv("ee_profile_all.csv")

print("Metadata shape:", meta.shape)
print("Profile shape:", profile.shape)
```

```python
print("Missing values in metadata:")
print(meta.isna().sum().sort_values(ascending=False).head(20))

print("\nMissing values in profile:")
print(profile.isna().sum().sort_values(ascending=False))
```

```python
print(
    "Duplicate employer references in metadata:",
    meta["ee_ref"].duplicated().sum()
)
```

```python
duplicate_keys = profile.duplicated(
    subset=["ee_ref", "submission_year", "occupation_level"]
)

print("Duplicate employer-year-occupation records:", duplicate_keys.sum())
```

```python
print(profile["occupation_level"].value_counts())
```

```python
profile["occupation_level"] = (
    profile["occupation_level"]
    .astype("string")
    .str.strip()
    .replace({
        "Temporary": "Temporary employees"
    })
)
```

```python
demographic_map = {
    "African_Female": ("African", "Female"),
    "AFRICAN_male": ("African", "Male"),
    "coloured_female": ("Coloured", "Female"),
    "coloured_MALE": ("Coloured", "Male"),
    "foreign_female": ("Foreign", "Female"),
    "foriegn_male": ("Foreign", "Male"),
    "Indian_FEMALE": ("Indian", "Female"),
    "indian_male": ("Indian", "Male"),
    "white_female": ("White", "Female"),
    "white_male": ("White", "Male")
}

print("Demographic mapping defined.")
```

```python
for column in meta.select_dtypes(include=["object"]).columns:
    meta[column] = meta[column].apply(
        lambda value: value.strip()
        if isinstance(value, str)
        else value
    )
```

```python
# Keep the employer reference as the identifier.
meta_long = meta.melt(
    id_vars=["ee_ref"],
    var_name="field_year",
    value_name="value"
)

# Separate the field name from the year.
meta_long[["attribute", "year"]] = (
    meta_long["field_year"]
    .str.extract(r"^(.*)_(\d{4})$")
)

# Convert the year to a numeric type.
meta_long["year"] = pd.to_numeric(
    meta_long["year"],
    errors="coerce"
)

# Remove the original combined field name.
meta_long = meta_long.drop(columns=["field_year"])

# Reshape the metadata into one employer-year record.
meta_long = meta_long.pivot(
    index=["ee_ref", "year"],
    columns="attribute",
    values="value"
).reset_index()

# Remove the columns index name.
meta_long.columns.name = None

print("Restructured metadata shape:", meta_long.shape)

display(meta_long.head())
```

```python
demographic_columns = list(demographic_map.keys())

print(demographic_columns)
```

```python
female_columns = [
    column for column, (_, gender) in demographic_map.items()
    if gender == "Female"
]

male_columns = [
    column for column, (_, gender) in demographic_map.items()
    if gender == "Male"
]

profile["female_total"] = profile[female_columns].sum(axis=1)

profile["male_total"] = profile[male_columns].sum(axis=1)

profile["total_employees"] = (
    profile["female_total"] + profile["male_total"]
)

display(
    profile[
        [
            "submission_year",
            "ee_ref",
            "occupation_level",
            "female_total",
            "male_total",
            "total_employees"
        ]
    ].head()
)
```

```python
print(
    "Missing demographic counts:",
    profile[demographic_columns].isna().sum().sum()
)

print(
    "Negative demographic counts:",
    (profile[demographic_columns] < 0).sum().sum()
)
```

```python
merged = profile.merge(
    meta_long,
    left_on=["ee_ref", "submission_year"],
    right_on=["ee_ref", "year"],
    how="left",
    indicator=True,
    validate="many_to_one"
)

print("Profile rows:", len(profile))
print("Merged rows:", len(merged))

print("\nMerge results:")
print(merged["_merge"].value_counts())
```

```python
merged = merged.drop(columns=["_merge", "year"])
```

```python
print(merged.shape)
```

```python
print(
    merged[
        [
            "dti_registration_name",
            "industry_sector",
            "business_type",
            "province_physical",
            "city_town_physical"
        ]
    ].isna().sum().sort_values(ascending=False)
)
```

```python
suspect = merged[
    merged["dti_registration_name"]
    .astype("string")
    .str.contains(
        "KWAZULU NATAL PROVINCIAL DEPARTMENT OF EDUCATION",
        case=False,
        na=False
    )
]

display(
    suspect[
        [
            "submission_year",
            "dti_registration_name",
            "industry_sector",
            "occupation_level",
            "total_employees"
        ]
    ].drop_duplicates()
)
```

```python
workforce_trends = (
    merged.groupby("submission_year")
    .agg(
        total_employees=("total_employees", "sum"),
        female_employees=("female_total", "sum"),
        male_employees=("male_total", "sum")
    )
    .reset_index()
)

workforce_trends["female_share_pct"] = (
    workforce_trends["female_employees"]
    / workforce_trends["total_employees"]
    * 100
)

workforce_trends["male_share_pct"] = (
    workforce_trends["male_employees"]
    / workforce_trends["total_employees"]
    * 100
)

display(workforce_trends.round(2))
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    workforce_trends["submission_year"],
    workforce_trends["female_share_pct"],
    marker="o",
    label="Female share"
)

plt.plot(
    workforce_trends["submission_year"],
    workforce_trends["male_share_pct"],
    marker="o",
    label="Male share"
)

plt.xlabel("Reporting year")
plt.ylabel("Share of reported workforce (%)")
plt.title("Gender Composition of the Reported Workforce, 2013–2024")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_management = merged[
    merged["occupation_level"].str.strip().eq("Top management")
].copy()

display(top_management.head())
```

```python
top_management_trend = (
    top_management.groupby("submission_year")
    .agg(
        female_employees=("female_total", "sum"),
        male_employees=("male_total", "sum"),
        total_employees=("total_employees", "sum")
    )
    .reset_index()
)

top_management_trend["female_share_pct"] = (
    top_management_trend["female_employees"]
    / top_management_trend["total_employees"]
    * 100
)

display(top_management_trend.round(2))
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    top_management_trend["submission_year"],
    top_management_trend["female_share_pct"],
    marker="o"
)

plt.xlabel("Reporting year")
plt.ylabel("Female share of top management (%)")
plt.title("Female Representation in Top Management, 2013–2024")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_management_2024 = merged[
    (merged["submission_year"] == 2024)
    & (merged["occupation_level"] == "Top management")
].copy()
```

```python
industry_analysis = (
    top_management_2024.groupby("industry_sector", dropna=False)
    .agg(
        employer_count=("ee_ref", "nunique"),
        female_employees=("female_total", "sum"),
        male_employees=("male_total", "sum"),
        total_employees=("total_employees", "sum")
    )
    .reset_index()
)

industry_analysis["female_share_pct"] = (
    industry_analysis["female_employees"]
    / industry_analysis["total_employees"]
    * 100
)

industry_analysis = industry_analysis.sort_values(
    "female_share_pct",
    ascending=True
)

display(industry_analysis.round(2))
```

```python
industry_analysis_10plus = industry_analysis[
    industry_analysis["employer_count"] >= 10
].copy()

display(industry_analysis_10plus.round(2))
```

```python
industry_analysis_10plus = industry_analysis[
    industry_analysis["employer_count"] >= 10
].copy()

display(industry_analysis_10plus.round(2))
```

```python
chart_data = industry_analysis_10plus.sort_values(
    "female_share_pct",
    ascending=True
)

plt.figure(figsize=(12, 7))

plt.barh(
    chart_data["industry_sector"].fillna("Missing industry"),
    chart_data["female_share_pct"]
)

plt.xlabel("Female share of top management (%)")
plt.ylabel("Industry")
plt.title(
    "Female Representation in Top Management by Industry, 2024\n"
    "(Industries with at least 10 employer references)"
)

plt.tight_layout()
plt.show()
```

```python
population_group_columns = {
    "African": ["African_Female", "AFRICAN_male"],
    "Coloured": ["coloured_female", "coloured_MALE"],
    "Foreign": ["foreign_female", "foriegn_male"],
    "Indian": ["Indian_FEMALE", "indian_male"],
    "White": ["white_female", "white_male"]
}

population_trends = []

for year, group in merged.groupby("submission_year"):
    total = group["total_employees"].sum()

    for population_group, columns in population_group_columns.items():
        count = group[columns].sum().sum()

        population_trends.append({
            "submission_year": year,
            "population_group": population_group,
            "employee_count": count,
            "share_pct": count / total * 100 if total else np.nan
        })

population_trends = pd.DataFrame(population_trends)

display(population_trends.round(2))
```
